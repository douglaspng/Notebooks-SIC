```python
from pathlib import Path
import pandas as pd
from IPython.display import display

print("Versão do pandas:", pd.__version__)
```

```python
url_dados = "https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/vaccinations/vaccinations.csv"
arquivo_local = Path("vaccinations.csv")

fonte = arquivo_local if arquivo_local.exists() else url_dados
df_vacinacao = pd.read_csv(fonte)
print(f"Fonte Utilizada: {fonte}")
print(f"Dimensões(linhas, colunas): {df_vacinacao.shape}")
```

```python
print(f"Primeiras 5 linhas")
display(df_vacinacao.head())

print("Tipos e Valoes não-nulos: ")
df_vacinacao.info()

print("Resumo das clnas numéricas: ")
display(df_vacinacao.describe())
```

```python
df_vacinacao['date'] = pd.to_datetime(df_vacinacao['date'], errors='coerce')
print(f"Datas inválidas: {df_vacinacao['date'].isna().sum()}")
df_vacinacao.set_index('date', inplace=True)
df_vacinacao.sort_index(inplace=True)

print(f"Localidades distintas, incluindo agregados: {df_vacinacao['location'].nunique()}")
print(f"Intervalo Observado: {df_vacinacao.index.min()}, até {df_vacinacao.index.max()}")
display(df_vacinacao.head())
```

```python
vacinacaoPorpais = df_vacinacao.groupby("location")

print(f"Quantidade de grupos: {len(vacinacaoPorpais)}")
print(f"Quantidade total de linhas: {len(df_vacinacao)}")
print(f"Cinco linhas por grupos produzem: {len(vacinacaoPorpais.head())}, linhas por conjunto")
print("Quantidade de registro nas primeiras cinco localidades: ")
display(vacinacaoPorpais.size().head())
```

```python
df_Brasil = df_vacinacao[df_vacinacao["location"] == "Brazil"].copy()
colunas = ["location", "total_vaccinations", "total_vaccinations_per_hundred"]
display(df_Brasil[colunas].head())

brasil_acima_100_milhoes = df_Brasil[df_Brasil["total_vaccinations"] > 100_000_000]
print(f"Registros do Brasil de 100 milhões de doses: {len(brasil_acima_100_milhoes)}")
display(brasil_acima_100_milhoes[colunas].head())
```

```python
top_vacinacao = df_vacinacao.sort_values(by="total_vaccinations", ascending=False)
display(top_vacinacao[["location", "total_vaccinations"]].head(10))
```

```python
from numpy._core.fromnumeric import sort
historico = df_vacinacao.reset_index()

df_localidades = historico[historico['iso_code'].str.len().eq(3)].copy()

df_ultimo = (
    df_localidades.dropna(subset=["date", "total_vaccinations"])
    .sort_values("date")
    .groupby("location", sort=False)
    .tail(1)
    .reset_index(drop=True)
)

top_10_paises = df_ultimo.sort_values(by="total_vaccinations", ascending=False).head(10)
print("Dez localidades do recorte com maior total de doses no último registro válido: ")
display(top_10_paises[["location", "date", "total_vaccinations"]])
```

```python
df_mediaPaises = (
    df_localidades.groupby("location")["total_vaccinations"]

    .mean().reset_index()

)

resultado_por_pais = df_mediaPaises[['location', 'total_vaccinations']]

print(resultado_por_pais.head(10))
```
