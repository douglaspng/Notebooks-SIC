```python
from datetime import datetime
import sys

diadoLote = datetime.now().strftime("%d:%m:%Y")


print("""

"Olá, sou o CQI (Controle de Qualidade Inteligente), estou aqui para monitorar e verificar todos os padrões de qualide de cada peça que irá passar pelo meu sistema.

Padrões que a peça precisa seguir.

    - Temperatura de cada objeto necessita estar em 20°C até 45°C;
    - Voltagem de cada peça necessita estar em 12V, com a tolerância a no máximo 13.5°V

""")

while True:
    checkdeInicializacao = input(f"Você deseja iniciar a verificação de qualidade do lote do dia {diadoLote}? ").lower()

    if checkdeInicializacao in ["sim", "s"]:
        print("Iniciando verificação de qualidade...")
        break
    elif checkdeInicializacao in ["não", "nao", "n"]:
        print("Encerando o programa.")
        sys.exit()
    else:
        print("Opção inválida! Tente novamente...\n")

print(f"""\n\n

        Iniciando a verificação de qualidade do lote - {diadoLote}

        INICIANDO A VERIFICAÇÃO DE TEMPERATURA

\n\n""")

pecascomDefeitos = 0
pecaTempValidadas = 0
pecasVoltValidadas = 0
pecasnosPadroes = 0
processoInterrompido = False

loteTamanho = int(input("\n\nIndique o tamanho do lote (Apenas números): \n"))

if loteTamanho > 0:
    for i in range(loteTamanho):
        temperatura = int(input("\nInsira a temperatura da peça (Apenas números): \n"))

            voltagem = float(input("Insira a voltagem da peça (Apenas números): \n"))
            if voltagem < 0:
                print("Voltagem inválida! Não pode ser negativa. Tente novamente.")
            else:
                break

        # Classificação da peça
        if 20 <= temperatura <= 45 and voltagem == 12:
            print("Peça APROVADA.")
            pecasAprovadas += 1
        elif 20 <= temperatura <= 45 and 12.1 <= voltagem <= 13.5:
            print("Peça APROVADA COM RESSALVA.")
            pecasAprovadasRessalva += 1
        else:
            print("Peça REJEITADA.")
            pecasRejeitadas += 1
            defeitosCriticos += 1

        # Regra: parada obrigatória da máquina ao atingir 5 defeitos críticos
        if defeitosCriticos == 5:
            print("\nLimite de 5 defeitos críticos atingido. Parada obrigatória da máquina.")
            processoInterrompido = True
            break

    if processoInterrompido:
        print(f"\n\nProcesso encerrado antecipadamente: limite de defeitos críticos atingido.")
    else:
        print(f"\n\nTodas as peças do lote foram inspecionadas com sucesso!")

    print(f"Total de peças aprovadas: {pecasAprovadas}")
    print(f"Total de peças aprovadas com ressalva: {pecasAprovadasRessalva}")
    print(f"Total de peças rejeitadas: {pecasRejeitadas}")
    print(f"Total de defeitos críticos: {defeitosCriticos}")
else:
    print("\nO tamanho do lote deve ser maior que zero para iniciar a inspeção.")
```
