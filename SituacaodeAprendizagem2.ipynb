```python
inventario = []


# Neste Código as funções foram criadas com a juda da IA (VibeCode)
# Funções de Validação de entrada

def validarEntrada(mensagem, tipo="texto"):
    """
    Função única de validação de entrada, reutilizada em todas as opções do menu.
    tipo="texto"   -> garante que o texto digitado não fique em branco.
    tipo="inteiro" -> garante que o valor digitado seja um inteiro >= 0.
    """
    while True:
        valor = input(mensagem).strip()

        if tipo == "texto":
            if len(valor) > 0:
                return valor
            print("Erro: este campo não pode ficar em branco.")

        elif tipo == "inteiro":
            try:
                valor = int(valor)
                if valor < 0:
                    print("Erro: o valor não pode ser negativo.")
                    continue
                return valor
            except ValueError:
                print("Erro: digite apenas números inteiros.")


def buscarFerramentaPorCodigo(codigo):
    """Percorre o inventario e retorna o dicionário da ferramenta, ou None se não achar."""
    for ferramenta in inventario:
        if ferramenta["codigo"] == codigo:
            return ferramenta
    return None


def cadastrarFerramenta():
    print("\n--- Cadastro de Ferramenta ---")

    codigo = validarEntrada("Código da ferramenta: ")

    # Validação avançada: verificar se o código já existe
    if buscarFerramentaPorCodigo(codigo) is not None:
        print(f"Erro: já existe uma ferramenta cadastrada com o código '{codigo}'.")
        return

    nome = validarEntrada("Nome da ferramenta: ")
    setor = validarEntrada("Setor da ferramenta: ")
    quantidade = validarEntrada("Quantidade em estoque: ", tipo="inteiro")
    statusCalibracao = validarEntrada("Status da calibração (ex: Aprovado/Reprovado): ")

    novaFerramenta = {
        "codigo": codigo,
        "nome": nome,
        "setor": setor,
        "quantidade": quantidade,
        "status_calibracao": statusCalibracao
    }

    inventario.append(novaFerramenta)
    print(f"\nFerramenta '{nome}' cadastrada com sucesso!")


def listarFerramentas():
    print("\n--- Lista de Ferramentas ---")

    if len(inventario) == 0:
        print("Nenhuma ferramenta cadastrada.")
        return

    for ferramenta in inventario:
        print(f"Código: {ferramenta['codigo']} | Nome: {ferramenta['nome']} | "
              f"Setor: {ferramenta['setor']} | Estoque: {ferramenta['quantidade']} | "
              f"Calibração: {ferramenta['status_calibracao']}")


def buscarFerramenta():
    print("\n--- Busca de Ferramenta ---")

    codigo = validarEntrada("Digite o código da ferramenta que deseja buscar: ")
    encontrado = False

    for ferramenta in inventario:
        if ferramenta["codigo"] == codigo:
            print(f"\nCódigo: {ferramenta['codigo']}")
            print(f"Nome: {ferramenta['nome']}")
            print(f"Setor: {ferramenta['setor']}")
            print(f"Estoque: {ferramenta['quantidade']}")
            print(f"Calibração: {ferramenta['status_calibracao']}")
            encontrado = True
            break

    if not encontrado:
        print(f"Ferramenta com código '{codigo}' não encontrada.")


def atualizarEstoque():
    print("\n--- Atualização de Estoque ---")

    codigo = validarEntrada("Digite o código da ferramenta que deseja atualizar: ")
    ferramenta = buscarFerramentaPorCodigo(codigo)

    if ferramenta is not None:
        print(f"Estoque atual de '{ferramenta['nome']}': {ferramenta['quantidade']}")
        novaQuantidade = validarEntrada("Digite a nova quantidade em estoque: ", tipo="inteiro")
        ferramenta["quantidade"] = novaQuantidade
        print("Estoque atualizado com sucesso!")
    else:
        print(f"Ferramenta com código '{codigo}' não encontrada.")


def excluirFerramenta():
    print("\n--- Exclusão de Ferramenta ---")

    codigo = validarEntrada("Digite o código da ferramenta que deseja excluir: ")
    ferramenta = buscarFerramentaPorCodigo(codigo)

    if ferramenta is not None:
        inventario.remove(ferramenta)
        print(f"Ferramenta '{ferramenta['nome']}' excluída com sucesso!")
    else:
        print(f"Ferramenta com código '{codigo}' não encontrada.")


def menuPrincipal():
    while True:
        print("""
========================================
   ALMOXARIFADO DE FERRAMENTAS - AeroTech
========================================
[1] Cadastrar ferramenta
[2] Listar ferramentas
[3] Buscar ferramenta
[4] Atualizar estoque
[5] Excluir ferramenta
[6] Sair
========================================""")

        opcao = input("Escolha uma opção: ").strip()

        if opcao == "1":
            cadastrarFerramenta()
        elif opcao == "2":
            listarFerramentas()
        elif opcao == "3":
            buscarFerramenta()
        elif opcao == "4":
            atualizarEstoque()
        elif opcao == "5":
            excluirFerramenta()
        elif opcao == "6":
            print("\nEncerrando o sistema. Até logo!")
            break
        else:
            print("\nOpção inválida! Escolha um número de 1 a 6.")



if __name__ == "__main__":
    menuPrincipal()
```
