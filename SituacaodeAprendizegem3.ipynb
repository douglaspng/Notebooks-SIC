```python
frota = {}

def validarInteiroNaoNegativo(mensagem):
    while True:
        valor = input(mensagem).strip()
        try:
            valor = int(valor)
            if valor < 0:
                print("Erro: o valor não pode ser negativo.")
                continue
            return valor
        except ValueError:
            print("Erro: digite apenas números inteiros.")

def cadastrarMaquina():
    print("\n--- Cadastro de Máquina ---")
    idMaquina = input("Digite o ID da máquina (ex: CNC-01): ").strip()

    if idMaquina in frota:
        print(f"Erro: a máquina '{idMaquina}' já está cadastrada.")
        return

    frota[idMaquina] = {
        "status": "Operando",
        "boas": 0,
        "refugadas": 0,
        "parada_min": 0
    }

    print(f"Máquina '{idMaquina}' cadastrada com sucesso! Status inicial: Operando.")

def registrarProducao():
    print("\n--- Registro de Produção ---")
    idMaquina = input("Digite o ID da máquina: ").strip()

    if idMaquina not in frota:
        print(f"Erro: a máquina '{idMaquina}' não está cadastrada.")
        return

    maquina = frota[idMaquina]

    if maquina["status"] != "Operando":
        print(f"Erro: a máquina '{idMaquina}' está PARADA. Não é possível registrar produção.")
        return

    boas = validarInteiroNaoNegativo("Quantidade de peças boas: ")
    refugadas = validarInteiroNaoNegativo("Quantidade de peças refugadas: ")

    maquina["boas"] += boas
    maquina["refugadas"] += refugadas

    print(f"Produção registrada! Total acumulado -> Boas: {maquina['boas']} | "
          f"Refugadas: {maquina['refugadas']}")

def registrarParada():
    print("\n--- Registro de Parada ---")
    idMaquina = input("Digite o ID da máquina: ").strip()

    if idMaquina not in frota:
        print(f"Erro: a máquina '{idMaquina}' não está cadastrada.")
        return

    maquina = frota[idMaquina]
    maquina["status"] = "Parada"

    minutosParados = validarInteiroNaoNegativo("Quantidade de minutos parados: ")
    maquina["parada_min"] += minutosParados

    print(f"Máquina '{idMaquina}' marcada como PARADA. "
          f"Total acumulado de parada: {maquina['parada_min']} min.")

def retomarProducao():
    print("\n--- Retomada de Produção ---")
    idMaquina = input("Digite o ID da máquina: ").strip()

    if idMaquina not in frota:
        print(f"Erro: a máquina '{idMaquina}' não está cadastrada.")
        return

    maquina = frota[idMaquina]
    maquina["status"] = "Operando"

    print(f"Máquina '{idMaquina}' voltou a OPERAR.")

def exibirDashboard():
    print("\n========== DASHBOARD DE PRODUÇÃO ==========")

    if len(frota) == 0:
        print("Nenhuma máquina cadastrada.")
        return

    for id_maq, dados in frota.items():
        total = dados["boas"] + dados["refugadas"]

        print(f"\nMáquina: {id_maq}")
        print(f"  Status: {dados['status']}")
        print(f"  Peças boas: {dados['boas']}")
        print(f"  Peças refugadas: {dados['refugadas']}")
        print(f"  Tempo parado: {dados['parada_min']} min")


        if total > 0:
            qualidade = (dados["boas"] / total) * 100
            print(f"  Índice de Qualidade: {qualidade:.2f}%")
        else:
            print("  Índice de Qualidade: sem produção registrada ainda.")

    print("\n=============================================")

def menuPrincipal():
    while True:
        print("""
========================================
   MONITORAMENTO CNC - TechSteel Usinagem
========================================
[1] Cadastrar máquina
[2] Registrar produção
[3] Registrar parada
[4] Retomar produção
[5] Dashboard
[6] Sair
========================================""")

        opcao = input("Escolha uma opção: ").strip()

        match opcao:
            case "1":
                cadastrarMaquina()
            case "2":
                registrarProducao()
            case "3":
                registrarParada()
            case "4":
                retomarProducao()
            case "5":
                exibirDashboard()
            case "6":
                print("\nEncerrando o sistema. Até logo!")
                break
            case _:
                print("\nOpção inválida!")

if __name__ == "__main__":
    menuPrincipal()
```
