#Pilhas
- Stack
- LIFO Last In, First Out: Uma pilha segue LIFO: o último item inserido é o primeiro removido.
Vamos simular o histórico de ações de um editor

```python
def empilhar(pilha, item):
  pilha.append(item)

def desempilhar(pilha):
  if not pilha:
    return None
  return pilha.pop()
```

```python
historico = []
empilhar(historico, "Digitou o Título")
empilhar(historico, "Alterou uma cor")
empilhar(historico, "Apagou uma palavra")

print(f"Histórico: {historico}")
```

```python
ultimaAcao = desempilhar(historico)

print(f"Ação Desfeita: {ultimaAcao} ")
print(f"Histórico Restante: {historico}")
print(f"A pilha está vazia? {not historico}")
```

#Desafio:
- Crie um histórico de páginas visitadas. Ao chamar voltar , a página atual deve ser removida e a anterior deve ser informada.

```python
def visitar(historico, pagina):
  historico.append(pagina)


def voltar(historico):

  if len(historico) >= 2:
    historico.pop()

    return historico[-1]

  else:
    return historico[0]
```

```python
paginas = []

visitar(paginas, "Inicio")
visitar(paginas, "Produtos")
visitar(paginas, "Carrinho")

print(voltar(paginas))
```

#Filas
- Queue
- FIFO - Firts In, Firts Out: o primeiro chamado recebido será o primeiro atendido.
Primeiro veremos uma lista; depois, usaremos deque.

```python
filaLista = []
filaLista.append("001 - Computador não Liga")
filaLista.append("002 - Problema de Rede")
filaLista.append("003 - Impressora não funciona")
```

```python
print(f"Fila: {filaLista}")
primeiro = filaLista.pop(0)
print(f"Atendido: {primeiro}")
print(f"Restante: {filaLista}")
```

```python
from collections import deque
```

```python
filaChamados = deque()
filaChamados.append("001 - Computador não Liga")
filaChamados.append("002 - Problema de Rede")
filaChamados.append("003 - Impressora não funciona")


if filaChamados:
  atendido = filaChamados.popleft()
  print(f"Atendendo: {atendido}")

print(f"Fila restante: {list(filaChamados)}")
```

```python
segundo = filaChamados.popleft() if filaChamados else None
print(f"Segundo atendimento: {segundo}")
print(f"Próximo da fila: {filaChamados[0] if filaChamados else 'Fila Vazia'}")
```

- Adicione dois chamados e crie um WHILE que atenda toda a fila. Antes de Executar, escreva a ordem prevista.

```python
filaChamados.append("001 — Computador não liga")
filaChamados.append("002 — Problema de rede")
filaChamados.append("003 — Impressora não funciona")

if filaChamados:
    atendido = filaChamados.popleft()
    print("Atendendo:", atendido)
    print("Fila restante:", list(filaChamados))

segundo = filaChamados.popleft() if filaChamados else None
print("Segundo atendimento:", segundo)
print("Próximo da fila:", filaChamados[0] if filaChamados else "Fila vazia")

filaChamados.append("004 — Teclado com defeito")
filaChamados.append("005 — Conta bloqueada")

print("\nFila antes de finalizar os atendimentos:")
print(list(filaChamados))

print("\n--- Atendendo os chamados restantes ---")

while filaChamados:
    atendido = filaChamados.popleft()
    print("Atendendo:", atendido)

print("\nFila final:", list(filaChamados))
```

- Situação problema: A equipe de patrimônio possui a lista [1020,1045,1088,1120,1156] e precisa localizar o patrimônio 1120. A lista não esta ordenada e não existe índice auxiliar.

```python
#A busca compara cada elemento até encontrar o alvo ou terminar a lista...

def pesquisa_sequencial(valores, procurado):
  for indice, valor in enumerate(valores):
    print(f"Comparando {procurado} com {valor}")

    if valor == procurado:
      return indice

  return -1

```

```python
patrimonios = [1020,1045,1088,1120,1156]
indice = pesquisa_sequencial(patrimonios,1120)

print(f"Indice encontrado: {indice}")
indice_inexistente = pesquisa_sequencial(patrimonios, 5763557)
print(f"Resultado ausente: {indice_inexistente}")
```

```python
#Teste de Passos

def pesquisa_sequencial_com_passos(valores, procurado):
  comparacoes = 0
  for indice, valor in enumerate(valores):
    comparacoes +=1
    print(f"Comparando {procurado} com {valor}")

    if valor == procurado:
      return indice, comparacoes


  return -1, comparacoes
```

```python
patrimonios = [1020,1045,1088,1120,1156]
print(pesquisa_sequencial_com_passos(patrimonios, 1023))
print(pesquisa_sequencial_com_passos(patrimonios, 1156))
print(pesquisa_sequencial_com_passos(patrimonios, 9999))
```

- Situação problema: Quando procuramos uma palavra em um dicionário físico, nornalmente abrimos perto do meio e decidimos se devemos seguir para as páginas anteriores ou posteriores. Não começamos pela primeira página.

```python
#Pesquisa Binária: a busca binária exige uma lista ordenada e elimina metade do intervalo a cada comparação

def pesquisa_binaria(valores_ordenados, procurado):
    inicio = 0
    fim = len(valores_ordenados) - 1
    comparacoes = 0
    while inicio <= fim:
        meio = (inicio + fim) // 2
        valor_do_meio = valores_ordenados[meio]
        comparacoes += 1
        print(
            f"início={inicio}, fim={fim}, "
            f"meio={meio}, valor={valor_do_meio}"
        )
        if valor_do_meio == procurado:
            return meio, comparacoes
        elif procurado < valor_do_meio:
            fim = meio - 1
        else:
            inicio = meio + 1
    return -1, comparacoes
```

```python
numeros = [10, 20, 30, 40, 50, 60, 70, 80, 90]
indice, comparacoes = pesquisa_binaria(numeros, 50)
print("Índice:", indice)
print("Comparações:", comparacoes)
```

```python
valores = list(range(1, 100_001))
alvo = 99_999
_, passos_lineares = pesquisa_sequencial_com_passos(valores, alvo)
_, passos_binarios = pesquisa_binaria(valores, alvo)
print("Pesquisa sequencial:", passos_lineares, "comparações")
print("Pesquisa binária:", passos_binarios, "comparações")
```

```python
#Tabela Hash
produtos = {
    1001:{"nome":"Teclado", "preco":150.00},
    1002:{"nome":"Mouse", "preco":110.00},
    1003:{"nome":"Monitor", "preco":986.70},
    1004:{"nome":"Capa", "preco":42.50},
    1005:{"nome":"Cabo", "preco":87.30},

}


```

```python
def buscarProduto(produtos, codigo):
  produto = produtos.get(codigo)

  if produto is None:
    return "Produto não encontrado!"
  return f"{produto['nome']} - R$ {produto['preco']:.2f}"
```

```python
print(buscarProduto(produtos, 1001))
print(buscarProduto(produtos, 31435468))
```
