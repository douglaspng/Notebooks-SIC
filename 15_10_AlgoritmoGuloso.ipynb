#Algoritmo Guloso(Greedy)

```python
def troco_guloso(valor, mostrar = False):
    escolhidas = []
    restante = valor

    for moeda in sorted(moeda, reverse = True):
        while moeda <= restante:
            escolhidas.append(moeda)
            restante -= moeda

            if mostrar:
                print(f"Pegue {moeda}; Falta devolver {restante}.")

    if restante !=0:
        return None
    return escolhidas
```

```python
resultado = troco_guloso(8, [5, 2, 1], mostrar = True)
print(f"Fichas escolhidas: {resultado}")
print(f"Quantidade de fichas: {len(resultado)}")
```

```python
moedas_especiais = [4, 3, 1]
guloso = troco_guloso(6, moedas_especiais, mostrar = True)
alternativa = [3, 3]

print(f"Resultado guloso: {guloso}, | moedas: {len(guloso)}")
print(f"Outra possibilidade: {alternativa}, | moedas: {len(alternativa)}")
print(f"Resultado guloso: {sum(guloso) == sum(alternativa) == 6}")
```
